# 🚀 Feature Engineering

In this notebook, temporal and geographical information will be converted into predictive variables for trip duration estimation.

```python
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
from src.config import CLEAN_DATA_PATH
from src.config import PROCESSED_DATA_PATH
from src.feature_engineering import (
    haversine_distance,
    manhattan_distance,
    calculate_bearing,
    create_datetime_features,
    encode_categorical_features,
    drop_unused_columns
)

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
## Load clean dataset

df = pd.read_csv(CLEAN_DATA_PATH)
```

```python
df = create_datetime_features(df)
```

```python
# harvesine distance

df["haversine_distance"] = haversine_distance(
    df["pickup_latitude"],
    df["pickup_longitude"],
    df["dropoff_latitude"],
    df["dropoff_longitude"]
)
```

```python
# manhattan distace
df["manhattan_distance"] = manhattan_distance(
    df["pickup_latitude"],
    df["pickup_longitude"],
    df["dropoff_latitude"],
    df["dropoff_longitude"]
)
```

```python
# Bearing
df["bearing"] = calculate_bearing(
    df["pickup_latitude"],
    df["pickup_longitude"],
    df["dropoff_latitude"],
    df["dropoff_longitude"]
)
```

```python
# Encode categorical features
df = encode_categorical_features(df)
```

```python
# Drop unnecessary columns
df = drop_unused_columns(df)
```

```python
## Save Dataset

df.to_csv(PROCESSED_DATA_PATH, index=False)

print(f"✅ Cleaned dataset saved to:\n{PROCESSED_DATA_PATH}")
```

## 📝 Key Insights

- Datetime variables were transformed into multiple temporal features.
- Geographic coordinates were converted into meaningful distance measures.
- Bearing captures the travel direction between pickup and dropoff locations.
- Categorical variables were encoded into numerical format.
- The dataset is now ready for model training.
