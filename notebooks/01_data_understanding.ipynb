# 🚖 NYC Taxi Trip Duration Prediction


## 📌 Problem Statement

Predict the duration of a New York City taxi trip based on trip details such as pickup and dropoff locations, pickup datetime, passenger count, and other available features.

This is a supervised machine learning regression problem where the target variable is **trip_duration**.
